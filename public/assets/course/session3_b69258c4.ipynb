```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('images.jfif')   
```

```python
def imgshow(img):
    cv2.imshow('Img' , img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()  
```

```python
imgshow(img)
```

```python
plt.imshow(img)
```

```python
img_color = cv2.cvtColor(img , cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_color)
```

```python
img_crop = img_color[:,100:600]
```

```python
plt.imshow(img_crop)
```

```python
img_resize = cv2.resize(img_crop , (100 , 100) , cv2.INTER_AREA)
```

```python
plt.imshow(img_resize)
```

```python
img_gray = cv2.cvtColor(img_crop , cv2.COLOR_RGB2GRAY)
```

```python
plt.imshow(img_gray)
```

```python
imgshow(img_gray)
```

```python
ret , img_threshold = cv2.threshold(img_gray , 127 , 255 , 0)
```

```python
imgshow(img_threshold)
```

```python
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
face_detection = face_cascade.detectMultiScale(img_gray)
```

```python
[[147,103,198,198] , [127,167,198,198]]
```

```python
face_detection
```

```python
x = face_detection[0][0]
y = face_detection[0][1]
x2 = x + face_detection[0][2]
y2 = y + face_detection[0][3]

xb = face_detection[1][0]
yb = face_detection[1][1]
xb2 = x + face_detection[1][2]
yb2 = y + face_detection[1][3]
```

```python
img_rectangle = cv2.rectangle(img_gray , (x,y) , (x2 , y2) , (255 , 255 , 255) , 3)
img_rectangle = cv2.rectangle(img_gray , (xb,yb) , (xb2 , yb2) , (255 , 255 , 255) , 3)
```

```python
imgshow(img_rectangle)
```
